```python
import pandas as pd

raw_df = pd.read_csv("house_sales.csv")
clean_outliers_df = pd.read_csv("cleaned_jordan_outliers.csv")

#clean_no_outliers_df = pd.read_csv("cleaned_house_sales.csv")
```

```python
def compare_distribution(col):
    raw_stats = raw_df[col].describe()
    clean_stats = clean_outliers_df[col].describe()

    raw_skew = raw_df[col].skew()
    clean_skew = clean_outliers_df[col].skew()

    comparison = pd.concat(
        [raw_stats, clean_stats],
        axis=1
    )

    comparison.columns = ['RAW', 'CLEAN']

    # Add skew row
    comparison.loc['skew'] = [raw_skew, clean_skew]

    print(f"\n===== {col.upper()} DISTRIBUTION =====")
    display(comparison)


key_vars = [
    'price',
    'sqft_living',
    'sqft_lot',
    'bedrooms',
    'bathrooms'
]

for col in key_vars:
    compare_distribution(col)
```

```python
bed_raw = raw_df['bedrooms'].value_counts(normalize=True).sort_index()
bed_clean = clean_outliers_df['bedrooms'].value_counts(normalize=True).sort_index()

bed_compare = pd.concat(
    [bed_raw, bed_clean],
    axis=1
).fillna(0)

bed_compare.columns = ['RAW', 'CLEAN']

display(bed_compare)

bath_raw = raw_df['bathrooms'].value_counts(normalize=True).sort_index()
bath_clean = clean_outliers_df['bathrooms'].value_counts(normalize=True).sort_index()

bath_compare = pd.concat(
    [bath_raw, bath_clean],
    axis=1
).fillna(0)

bath_compare.columns = ['RAW', 'CLEAN']

display(bath_compare)
```
