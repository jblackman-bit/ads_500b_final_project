## Load Packages and Inspect Data
<a id='setup'></a>

```python
# Import the necessary packages
import numpy as np
import pandas as pd
import matplotlib.gridspec as gridspec
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
# Import the dataset using pandas read_csv
df = pd.read_csv("cleaned_house_sales.csv")

# Bring in data frame with outliers
df_with_outliers = pd.read_csv("cleaned_house_sales_with_outliers.csv")

# Ensure head displays all columns and no scientific notation
pd.set_option("display.max_columns", None)
pd.options.display.float_format = "{:,.2f}".format

# Display the top 5 rows
df.head()
```

```python
# Display the shape of the dataframe
shape_df = pd.DataFrame(
    df.shape,
    index=["Rows", "Columns"],
    columns=["Count"],
)
display(shape_df)

# Display count by data type
dtype_df = df.dtypes.value_counts().to_frame(name="Number of Columns")
dtype_df.index.name = "Data Type"
display(dtype_df.reset_index())
```

```python
# Display column number, column name, non-null count, and data type
info_df = pd.DataFrame(
    {
        "Column": df.columns,
        "Non-Null Count": df.notnull().sum().values,
        "Dtype": df.dtypes.values,
    }
)

display(info_df.reset_index(drop=True))
```

## Descriptive Statistics
<a id='stats'></a>

```python
# Calculate descriptive statistics including skewness and kurtosis
stats = df.describe().T
stats["skew"] = df.select_dtypes(include=[np.number]).skew()
stats["kurtosis"] = df.select_dtypes(include=[np.number]).kurt()

print("--- Descriptive Statistics with Skewness ---")
display(stats[["mean", "50%", "std", "min", "max", "skew", "kurtosis"]])
```

## Univariate Analysis
<a id='univariate'></a>

- Visually inspect distributions to assess normality. Logged variables exhibit more symmetric, near-normal distributions, which are better suited for linear regression.
- Removing extreme outliers further improves distributional shape and reduces the influence of extreme values.

```python
# Columns to show pre and post cleaning distributions
cols = ["price", "log_price"]

# Set up list for stats comparison
all_stats = []

for col in cols:
    # Calculate stats for both dataframes
    s_outliers = df_with_outliers[col].describe().to_frame(
        name=f"{col} (With Outliers)"
    )
    s_trim = df[col].describe().to_frame(name=f"{col} (Trimmed)")

    # Add skew and kurtosis
    for df_temp, s_temp in [
        (df_with_outliers, s_outliers),
        (df, s_trim),
    ]:
        s_temp.loc["skew"] = df_temp[col].skew()
        s_temp.loc["kurtosis"] = df_temp[col].kurt()

    all_stats.extend([s_outliers, s_trim])

    # Create visualizations for this specific column
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle(
        f'Analysis for {col.replace("_", " ").title()}',
        fontsize=16,
    )

    # Histogram: With Outliers
    sns.histplot(
        df_with_outliers[col],
        kde=True,
        ax=axes[0, 0],
        color="skyblue",
    )
    axes[0, 0].set_title("Distribution: With Outliers")

    # Histogram: Trimmed
    sns.histplot(df[col], kde=True, ax=axes[0, 1], color="steelblue")
    axes[0, 1].set_title("Distribution: Trimmed (Sans Outliers)")

    # Boxplot: With Outliers
    sns.boxplot(
        x=df_with_outliers[col],
        ax=axes[1, 0],
        color="lightgreen",
        fliersize=3,
    )
    axes[1, 0].set_title("Boxplot: With Outliers")

    # Boxplot: Trimmed
    sns.boxplot(
        x=df[col],
        ax=axes[1, 1],
        color="forestgreen",
        fliersize=3,
    )
    axes[1, 1].set_title("Boxplot: Trimmed")

    # Apply formatting to all axes in this figure
    for ax in axes.flat:
        ax.xaxis.set_major_formatter(ticker.EngFormatter())

print("--- Summary Comparison Table ---")
display(pd.concat(all_stats, axis=1))
```

```python
# Plot distribution and boxplots for all numeric variables

numeric_df = df.select_dtypes(include=["number"])
cols = [c for c in numeric_df.columns if c.lower() != "id"]

fig, axes = plt.subplots(len(cols), 2, figsize=(14, 4 * len(cols)))

for i, col in enumerate(cols):
    # Make the histogram
    sns.histplot(df[col], kde=True, ax=axes[i, 0], color="skyblue")
    axes[i, 0].set_title(f"Distribution of {col}")

    # Format X axis
    axes[i, 0].xaxis.set_major_formatter(ticker.EngFormatter())

    # Create lines for mean/median
    axes[i, 0].axvline(
        df[col].mean(),
        color="red",
        linestyle="--",
        label="Mean",
    )
    axes[i, 0].axvline(
        df[col].median(),
        color="blue",
        linestyle="--",
        label="Median",
    )
    axes[i, 0].legend()

    # Make the boxplot
    sns.boxplot(
        x=df[col],
        ax=axes[i, 1],
        color="lightgreen",
        fliersize=3,
    )
    axes[i, 1].set_title(f"Boxplot of {col}")
    axes[i, 1].xaxis.set_major_formatter(ticker.EngFormatter())

    # Make sure both plots show the same range
    xlims = axes[i, 0].get_xlim()
    axes[i, 1].set_xlim(xlims)

    sns.despine(ax=axes[i, 0])
    sns.despine(ax=axes[i, 1])

plt.tight_layout()
plt.show()
```

## Correlation
<a id='correl'></a>

1. Review the entire correlation matrix
2. Review the high positively correlated variables
3. Review the Lot correlation values
4. Review the high negatively correlated variables

```python
# plot total correlation matrix
corr_matrix = df.corr(numeric_only=True).round(2)

price_corr_sorted = corr_matrix["price"].sort_values(ascending=False)
corr_matrix_sorted = corr_matrix.loc[
    price_corr_sorted.index,
    price_corr_sorted.index,
].round(2)

mask = np.triu(np.ones_like(corr_matrix_sorted, dtype=bool))

plt.figure(figsize=(15, 15))
sns.heatmap(
    corr_matrix_sorted,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    mask=mask,
)
plt.title("Correlation Heatmap of Cleaned Data Set")
plt.show()
```

The dependent variable is log_price. Independent variables are treated as predictors of price rather than outcomes, reflecting the assumption that property characteristics influence price rather than the reverse.

```python
# plot the most positive values
pos_cols_to_keep = corr_matrix_sorted.columns[
    corr_matrix_sorted["price"] > 0.3
]
full_pos_context_df = corr_matrix_sorted.loc[
    pos_cols_to_keep,
    pos_cols_to_keep,
]
mask = np.triu(np.ones_like(full_pos_context_df, dtype=bool))

plt.figure(figsize=(14, 10))
sns.heatmap(
    full_pos_context_df,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    mask=mask,
)

plt.title("Strong to Moderate Positive Correlations")
plt.show()
```

Price and log_price are highly correlated; however, log_price is considered as a potential dependent variable to improve model fit and linearity. Log-transformed features, particularly log_sqft_living and log_sqft_above, show stronger relationships with log_price than with price.

Grade, sqft_living, sqft_above, and related variables exhibit strong correlations with one another, indicating potential multicollinearity. Model selection will require selecting among these features to avoid redundancy.

Bedrooms and bathrooms show moderate correlations with price and log_price and are strongly correlated with living space variables, reflecting underlying structural relationships.

View is moderately correlated with price and log_price but shows limited correlation with other predictors, suggesting it may contribute independent explanatory value.

```python
# plot the lot values that are highly correlated with one another
lot_cols = [col for col in corr_matrix_sorted.columns if "lot" in col]
selection = ["price"] + lot_cols

lot_matrix = corr_matrix_sorted.loc[selection, selection]
lot_matrix = lot_matrix.drop(
    columns=["living_lot_ratio"],
    index=["living_lot_ratio"],
    errors="ignore",
)

mask = np.triu(np.ones_like(lot_matrix, dtype=bool))

plt.figure(figsize=(10, 8))
sns.heatmap(
    lot_matrix,
    annot=True,
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    mask=mask,
)

plt.title("Price and Lot Correlation")
plt.show()
```

Lot-related variables are moderately to highly correlated with one another but show weak correlation with price, suggesting limited direct predictive value.

```python
# Plot the highly negative correlation values
original_cols = [
    "price",
    "grade",
    "sqft_living",
    "log_sqft_living",
    "sqft_living15",
    "sqft_above",
    "log_sqft_above",
    "bathrooms",
    "floors",
    "living_lot_ratio",
    "log_sqft_lot",
    "log_sqft_lot15",
    "house_age",
]

rows_to_keep = [
    "log_sqft_lot",
    "log_sqft_lot15",
    "house_age",
    "house_age_c",
]

filtered_df = corr_matrix_sorted.loc[rows_to_keep, original_cols]
rect_mask = pd.DataFrame(
    False,
    index=filtered_df.index,
    columns=filtered_df.columns,
)

for r in filtered_df.index:
    if r in filtered_df.columns:
        col_idx = filtered_df.columns.get_loc(r)
        rect_mask.loc[r, filtered_df.columns[col_idx:]] = True

plt.figure(figsize=(15, 5))
sns.heatmap(
    filtered_df,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    mask=rect_mask,
)

plt.title("High and Moderate Negative Correlations")
plt.show()
```

House age exhibits moderate negative correlations with structural features such as grade, living space, and bathrooms, reflecting differences in construction standards over time.

The living_lot_ratio shows a strong negative relationship with logged lot size variables, consistent with its definition. The relationship is more pronounced in log-transformed variables, indicating a clearer log-log relationship.

Although negatively correlated variables show weaker relationships with price, they are still considered for inclusion to evaluate their contribution to overall model performance.

## Bivariate Analysis
<a id='bivariate'></a>

- Evaluate relationships between price, log_price, and each potential predictor to assess linearity and model suitability.

```python
# Plot joint plots for all variables and price/log_price
top_features = df.select_dtypes(include=["number"]).columns
features_to_plot = [
    f for f in top_features if f not in ["price", "log_price", "id"]
]

for feature in features_to_plot:
    fig = plt.figure(figsize=(16, 6))
    gs = gridspec.GridSpec(1, 2)

    targets = ["price", "log_price"]

    for i, target in enumerate(targets):
        ax = fig.add_subplot(gs[i])

        sns.regplot(
            data=df,
            x=feature,
            y=target,
            ax=ax,
            scatter_kws={"alpha": 0.2},
            line_kws={"color": "red"},
        )

        ax.set_title(f"{feature} vs {target.capitalize()}")

    plt.suptitle(
        f"Correlation Comparison: {feature} vs Price and Log Price",
        y=1.05,
        fontsize=14,
    )
    plt.tight_layout()
    plt.show()
```
